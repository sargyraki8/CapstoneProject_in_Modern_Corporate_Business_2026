# stg_address

Runs the corresponding staging script to create or refresh the `stg_address` table in BigQuery.

```python
from pathlib import Path
import runpy

candidate_paths = [Path.cwd(), *Path.cwd().parents]
repo_root = next((path for path in candidate_paths if (path / "scripts").exists()), None)
if repo_root is None:
    raise FileNotFoundError("Could not locate the repository root containing the scripts directory.")
script_path = repo_root / "scripts" / "staging" / "stg_address.py"
runpy.run_path(str(script_path), run_name="__main__")
```
